```python
#Q2 b
def most_frequent(a):
    vals, counts = np.unique(a, return_counts=True)
    top = counts.max()
    for v in vals[counts == top]:
        print("value:", v, "count:", top, "indices:", np.where(a == v)[0])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("x")
most_frequent(x)
print("y")
most_frequent(y)
```

```python
#Q4
Pehal = np.linspace(10, 100, 25)
print(Pehal)
print("dimensions:", Pehal.ndim)
print("shape:", Pehal.shape)
print("total elements:", Pehal.size)
print("data type:", Pehal.dtype)
print("bytes:", Pehal.nbytes)

col = Pehal.reshape(25, 1)
print(col.shape)
print(Pehal.T.shape)
print(np.array_equal(Pehal.T, Pehal))
print(Pehal.reshape(1, 25).T.shape)
print("T on a 1D array does nothing, it needs 2D (reshape first) to actually transpose")
```

```python
#Q5
ucs420_pehal = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print(ucs420_pehal)
print("mean:", ucs420_pehal.mean())
print("median:", np.median(ucs420_pehal))
print("max:", ucs420_pehal.max())
print("min:", ucs420_pehal.min())
print("unique:", np.unique(ucs420_pehal))

reshaped_ucs420_pehal = ucs420_pehal.reshape(4, 3)
print(reshaped_ucs420_pehal)

resized_ucs420_pehal = np.resize(ucs420_pehal, (2, 3))
print(resized_ucs420_pehal)
```
