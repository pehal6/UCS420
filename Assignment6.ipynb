```python
#Q4
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```

```python
#Q4 a
print(X.shape)
print(X.ndim)
```

```python
#Q4 b
print(X.T)
print("transpose swaps rows and columns, so now each row is one feature (sleep, activity, stress) across all users")
```

```python
#Q4 c
XtX = X.T @ X
print(XtX)
```

```python
#Q4 d
print(np.linalg.det(XtX))
XtX_inv = np.linalg.inv(XtX)
print(XtX_inv)
```

```python
#Q4 e
beta = XtX_inv @ X.T @ y
print(beta)
```

```python
#Q4 f
print("each coefficient is the change in assistance score for one unit increase in that feature, keeping the others fixed")
print("sleep:", round(beta[0], 3), "activity:", round(beta[1], 3), "stress:", round(beta[2], 3))
```

```python
#Q4 g
new_user = np.array([5, 40, 7])
pred = new_user @ beta
print(pred)
```
