```python
#2(i)
scores = (10, 0, 20, 40, 10, 70, 0, 10)
print("Scores =", scores)

highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest appears:", lowest_count)
```

```python
#2(ii)
reversed_scores = list(reversed(scores))
print("Reversed tuple as list:", reversed_scores)
```

```python
#2(iii)

score = int(input("Enter a score: "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score is not present in the tuple.")
```

```python
#2(iv)

scores[0] = 100
#This error occurs because tuples are immutable.
```

```python
#2(v)

first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
#3(i)
import random
random.seed(1024170103)
numbers = [random.randint(100, 900) for i in range(100)]
print(numbers)
```

```python
#3(ii)

odd_numbers = [x for x in numbers if x % 2 != 0]

print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

```python
#3(iii)
even_numbers = [x for x in numbers if x % 2 == 0]
print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
```

```python
#3(iv)
prime_numbers = []
for num in numbers:
    if num > 1:
        prime = True

        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
        if prime:
            prime_numbers.append(num)
print("Prime numbers:", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

prime_numbers_comprehension = [
    num for num in numbers
    if num > 1 and all(num % i != 0 for i in range(2, int(num ** 0.5) + 1))
]
print("Prime numbers using list comprehension:", prime_numbers_comprehension)
```

```python
#3(v)
from collections import Counter
count = Counter(numbers)
most_common_number = count.most_common(1)[0][0]
frequency = count.most_common(1)[0][1]
print("Most frequent number:", most_common_number)
print("Number of times it occurs:", frequency)
```

```python
#4
digits = [1, 0, 2, 4, 1, 7, 0, 1, 0, 3]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)
```

```python
#4(i)
print("Union:", A.union(B))
```

```python
#4(ii)
print("Intersection:", A.intersection(B))
```

```python
#4(iii)
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

```python
#4(iv)
print("Symmetric difference:", A.symmetric_difference(B))
#gives elements present in either set but not both.
```

```python
#4(v)
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
#4(vi)
X = int(input("Enter a value to remove from A: "))
A.discard(X)
print("Set A after discard:", A)
# discard() is safer than remove() because it does not give an error
# if the value is not present in the set.
```

```python
#5
my_dict = {
    "name": "Pehal",
    "roll_no": "1024170103",
    "branch": "CSE",
    "age": 19,
    "city": "Khanna"
}
print(my_dict)
```

```python
#5(i)
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

```python
#5(ii)
my_dict["cgpa"] = 8
print(my_dict)
```

```python
#5(iii)
my_dict["age"] = my_dict["age"] + 1
print(my_dict)
```

```python
#5(iv)
dict1 = my_dict.copy()
dict2 = my_dict.copy()
# Using pop()
removed_value = dict1.pop("branch")
print("After pop():", dict1)
print("Returned value:", removed_value)
# Using del
del dict2["branch"]
print("After del:", dict2)
#pop() removes the key and returns its value.
#del removes the key but does not return its value.
```

```python
#5(v)
for key, value in my_dict.items():
    print(key, "->", value)
```

```python
#5(vi)
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not present in the dictionary.")
```

```python
#5(vii)
friend_dict = {
    "name": "P",
    "roll_no": "1024170104",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}
merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:")
print(merged_dict)
```

```python
#5(viii)
string_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}
print("String values dictionary:")
print(string_dict)
```
