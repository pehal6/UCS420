```python
#6
emp_data = {
    'Employee_ID': [101,102,103,104,105],
    'Name': ['Alice','Bob','Charlie','Diana','Edward'],
    'Department': ['HR','IT','IT','Marketing','Sales'],
    'Age': [29,34,41,28,38],
    'Salary': [50000,70000,65000,55000,60000],
    'Years_of_Experience': [4,8,10,3,12],
    'Joining_Date': ['2020-03-15','2017-07-19','2013-06-01','2021-02-10','2010-11-25'],
    'Gender': ['Female','Male','Male','Female','Male'],
    'Bonus': [5000,7000,6000,4500,5000],
    'Rating': [4.5,4.0,3.8,4.7,3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv('employees.csv', index=False)
emp_df
```

```python
#(e)
avg_salary = emp_df['Salary'].mean()
total_bonus = emp_df['Bonus'].sum()
youngest_age = emp_df['Age'].min()
highest_rating = emp_df['Rating'].max()

print("average salary is", avg_salary)
print("total bonus paid is", total_bonus)
print("youngest employee age is", youngest_age)
print("highest rating is", highest_rating)
```

```python
#(f)
emp_df.sort_values(by='Salary', ascending=False)
```

```python
#(g)
def performance(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

emp_df['Performance'] = emp_df['Rating'].apply(performance)
emp_df
```

```python
#(h)
emp_df.isnull().sum()
```

```python
#(i)
emp_df = emp_df.rename(columns={'Employee_ID':'ID'})
emp_df
```

```python
#(j)
exp_more_than_5 = emp_df[emp_df['Years_of_Experience'] > 5]
it_dept = emp_df[emp_df['Department'] == 'IT']

print(exp_more_than_5)
print(it_dept)
```

```python
#(k)
emp_df['Tax'] = emp_df['Salary'] * 0.10
emp_df
```
